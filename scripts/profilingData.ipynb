```python
import json
import os

import openpyxl


def count_specifications_and_measure(json_dir, excel_file):

    # Load the Excel workbook
    workbook = openpyxl.load_workbook(excel_file)


    # Dictionary to store the counts
    name_counts = {}
    value_counts = {}
    measure_counts = {}
    
    # Iterate through each JSON file in the directory
    for filename in os.listdir(json_dir):
        print(f"Processing {filename}")
        if filename.endswith(".json"):
            filepath = os.path.join(json_dir, filename)
            with open(filepath, "r") as f:
                try:
                    data = json.load(f)
                    for key, item in data.items():
                        if "Specifications" in item:
                            specifications = item["Specifications"]
                            for key, spec in specifications.items():
                                if "name" in spec:
                                    name = spec["name"].lower()
                                    name_counts[name] = name_counts.get(name, 0) + 1
                                if "value" in spec:
                                    value = spec["value"].lower()
                                    value_counts[value] = value_counts.get(value, 0) + 1            
                        if "Measure" in item:
                            measure = item["Measure"].lower()
                            measure_counts[measure] = measure_counts.get(measure, 0) + 1
                except Exception as e:
                    print(f"Skipping invalid JSON file: {filename} with error: {e}")

    counts = [name_counts, value_counts, measure_counts]
    # Write the counts to the Excel sheet
    for i in range(3):
        row = 2  # Start from row 2
        for key, count in counts[i].items():
            workbook.worksheets[i].cell(row=row, column=1).value = key
            workbook.worksheets[i].cell(row=row, column=2).value = count
            row += 1

    # Save the Excel file
    workbook.save(excel_file)


json_dir = "../fabio_misino_ortega_claims"
excel_file = "../NAME_PROFILING.xlsx"

count_specifications_and_measure(json_dir, excel_file)
```
